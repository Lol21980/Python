## 4. Chtenie i opisanie sreza

```python
import pandas as pd
df = pd.read_csv('sample_200k.csv', sep=';', encoding='utf-8-sig', low_memory=False)
print(df.shape)
display(df.dtypes.to_frame('tip'))
display((df.isna().sum()/len(df)*100).round(2).to_frame('propuski_%'))
```

## 5. Grafik: raspredelenie chislovogo stolbtsa / chislo zapisey po godam

```python
import matplotlib.pyplot as plt
df['Год'].value_counts().sort_index().plot(kind='bar', figsize=(10,3))
plt.xlabel('god'); plt.ylabel('chislo zapisey'); plt.title('Zapisey po godam (srez)')
plt.show()
```
