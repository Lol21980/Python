```python
import pandas as pd
pd.set_option('display.max_columns', 50)
SRC = r'C:\Users\artur\Desktop\датасеты\Земский\csv\1_датасет_население_возраст_и_пол.csv'
df = pd.read_csv(SRC, sep=';', encoding='utf-8-sig', low_memory=False)
print(df.shape)
display(df.head(3))
```

```python
display(df.dtypes.to_frame('tip'))
print('strok, stolbtsov:', df.shape)
print('unikalnykh stran:', df['Код страны'].nunique())
print('unikalnykh pokazateley:', df['Код показателя'].nunique())
print('gody:', int(df['Год'].min()), '-', int(df['Год'].max()))
```

```python
# dolya propuskov po stolbtsam
display((df.isna().sum()/len(df)*100).round(2).to_frame('propuski_%'))
# pustaya Snoska pochti vezde; Region/Income otsutstvuyut u agregatov
display(df['Код показателя'].value_counts().head(10))
```

```python
# skrytye defekty: znacheniya, ne razbirayushchiesya v chislo (porcha Excel-daty)
import re
month_re = re.compile(r'^(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)\.', re.IGNORECASE)
s = df['Значение'].astype(str)
mask = s.str.match(month_re, na=False)
print('pokhozhikh na daty-mesyats:', int(mask.sum()), f"({mask.mean()*100:.1f}%)")
display(s[mask].value_counts().head(10))
```

```python
# raspredelenie po godam (odin grafik)
import matplotlib.pyplot as plt
df['Год'].hist(bins=65)
plt.xlabel('god'); plt.ylabel('chislo zapisey'); plt.title('Raspredelenie zapisey po godam')
plt.show()
```
